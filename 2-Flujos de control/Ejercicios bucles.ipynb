```python

lista_numeros = []
lista_numeros_positivos = []

numero_str = input("Introduce un número, 0 para resolver la suma: ")
numero_entero = int(numero_str)

while numero_entero != 0:
    
    if numero_entero > 0:
        lista_numeros_positivos.append(numero_entero)
        print(sum(lista_numeros_positivos))
    
    else:
        lista_numeros.append(numero_entero)

    numero_str = input("Introduce un número, 0 para parar:")
    numero_entero = int(numero_str)

print("lista números positivos:",lista_numeros_positivos)
print("lista de todos los números:", lista_numeros+lista_numeros_positivos)

        



```

## Ejercicio 11
Suponiendo que siempre se cumple que las dos listas del ejercicio son iguales de tamaño, implementa un programa que calcule el ratio ingreso/gasto de cada elemento, y lo guarde en una nueva lista. Si hubiese algún problema con los datos, hay que almacenar un mensaje de error en la lista del resultado. Usa `try/except`
> ingresos = [100, 200, 500, 100, 600]
>
> gastos = [50, 20, 70, 0, 25]

```python
ratio_gasto_ingreso = []
ingresos = [100,200,500,100,600]
gastos = [50,20,70,0,25]


for i in range(len(ingresos)):
    try:
        ratio_gasto.append(ingresos[i]/gastos[i])

    except ZeroDivisionError:
        print("No se puede dividir por 0")
        ratio_gasto_ingreso.append("Zero division error")
    except IndexError:
        print("IndexError")
        ratio_gasto_ingreso.append("Index error")        
    except TypeError:
        print("TypeError")
        ratio_gasto_ingreso.append("Type error")

print(ratio_gasto_ingresos)

```
