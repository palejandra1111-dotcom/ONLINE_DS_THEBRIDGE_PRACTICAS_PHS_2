## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

````python
def conver_grad_rad():
    from math import pi
    grados = float(input("Por favor introduzca los grados: "))
    conversion = grados * (pi / 180)

    return conversion

conver_grad_rad()
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python

def paralelo():
    base = float(input("Introduzca la base de su paralelogramo: "))
    altura = float(input("Introduzca la altura de su paralelogramo: "))
    area = base * altura

    print(area)

paralelo()
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python

def usuario():
    a = input("Introduzca su nombre: ")
    b = input("Introduzca su apellido: ")
    c = input("Que edad tienes: ")

    if a == b  and b == c and c == a: 
        print("Todos son iguales")
    elif a == b or a == c or b == c:
        print("Al menos 2 son iguales")
    else:
        print("Ninguna son iguales")

usuario()
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10

````python
def suma_comparativa():
    a = float(input("introduzca el primer número: "))
    b = float(input("Introduzca el segundo número: "))
    suma = a + b

    if suma == 10:
        print("Es igual a 10")
    elif suma < 10:
        print("Es inferior a 10")
    elif suma > 10:
        print("Es superior a 10")

suma_comparativa()
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
quijote = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
mayus = quijote.upper()
print(mayus)
minus = quijote.lower()
print(minus)
titulo = quijote.title()
print(titulo)
dividido = quijote.split()
print(dividido)
otro_quijote = quijote.replace(",",";")
print(otro_quijote)
sin_a = quijote.replace("a","")
print(sin_a)

````
